```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent

sys.path.insert(
    0,
    str(ROOT / "src")
)

import pandas as pd

from simplexid.data import (
    load_terma,
    load_hoasa,
    prepare_terma,
    prepare_hoasa,
)
```

Prepare Terma Dataset

```python
terma_raw = load_terma()

terma_raw
```

```python
print(
    terma_raw["train"].features
)

print(
    terma_raw["train"][0]
)
```

```python
terma = prepare_terma(
    terma_raw
)

terma.head()
```

```python
terma.shape
```

```python
terma[
    terma["labels_available"]
][
    [
        "sentence_id",
        "sentence",
        "aspects",
        "opinions",
        "n_aspects",
        "n_opinions",
    ]
].head(20)
```

```python
terma[
    (
        terma["n_aspects"].fillna(0) > 1
    )
    |
    (
        terma["n_opinions"].fillna(0) > 1
    )
][
    [
        "sentence",
        "aspects",
        "opinions"
    ]
].head(20)
```

```python
labeled_terma = terma[
    terma["labels_available"]
]

print(
    "Labeled sentences:",
    len(labeled_terma)
)

print(
    "1 aspect + 1 opinion:",
    (
        (labeled_terma["n_aspects"] == 1)
        &
        (labeled_terma["n_opinions"] == 1)
    ).sum()
)

print(
    "Multi aspect:",
    (
        labeled_terma["n_aspects"] > 1
    ).sum()
)

print(
    "Multi opinion:",
    (
        labeled_terma["n_opinions"] > 1
    ).sum()
)

print(
    "No aspect:",
    (
        labeled_terma["n_aspects"] == 0
    ).sum()
)

print(
    "No opinion:",
    (
        labeled_terma["n_opinions"] == 0
    ).sum()
)
```

```python
# ============================================================
# FINAL CLEANUP TERMA
# ============================================================

terma["n_aspects"] = terma["n_aspects"].astype("Int64")
terma["n_opinions"] = terma["n_opinions"].astype("Int64")

terma.head()
```

```python
PROCESSED = (
    ROOT / "data" / "processed"
)

PROCESSED.mkdir(
    parents=True,
    exist_ok=True
)

terma.to_csv(
    PROCESSED / "terma_processed.csv",
    index=False
)
```

```python
terma_meta = terma[
    [
        "sentence_id",
        "dataset",
        "source_split",
        "labels_available",
        "n_aspects",
        "n_opinions",
    ]
].copy()
```

Processeing Hoasa Dataset

```python
hoasa_raw = load_hoasa()

hoasa = prepare_hoasa(
    hoasa_raw
)

hoasa.head()
```

```python
hoasa.shape
```

```python
hoasa[
    hoasa["source_split"] == "train"
].head(10)
```

```python
hoasa.to_csv(
    PROCESSED / "hoasa_processed.csv",
    index=False
)
```

```python
hoasa_meta = hoasa[
    [
        "sentence_id",
        "dataset",
        "source_split",
        "labels_available",
    ]
].copy()

hoasa_meta["n_aspects"] = None
hoasa_meta["n_opinions"] = None
```

```python
metadata = pd.concat(
    [
        terma_meta,
        hoasa_meta
    ],
    ignore_index=True
)

metadata.to_csv(
    PROCESSED
    / "sentence_metadata.csv",
    index=False
)
```
